Iterative function

```python
import time

# Iterative function
def sum_iterative(n):
    total = 0

    for i in range(1, n + 1):
        total += i

    return total


# User input
n = int(input("Enter a number: "))

# Measure execution time
start_time = time.perf_counter()

result = sum_iterative(n)

end_time = time.perf_counter()

# Display results
print("\nResult:", result)
print("Execution Time:", end_time - start_time, "seconds")
print("Time Complexity: O(n)")
print("Space Complexity: O(1)")
```

recursive function

```python
import time

# Recursive function
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


# User input
n = int(input("Enter a non-negative integer: "))

if n < 0:
    print("Please enter a non-negative integer.")
else:
    # Measure execution time
    start_time = time.perf_counter()

    result = factorial(n)

    end_time = time.perf_counter()

    # Display results
    print("\nFactorial:", result)
    print("Execution Time:", end_time - start_time, "seconds")
    print("Time Complexity: O(n)")
    print("Space Complexity: O(n)")
```

Iterative functions use loops and are generally faster and more memory-efficient.
Recursive functions call themselves and can make complex problems easier to
